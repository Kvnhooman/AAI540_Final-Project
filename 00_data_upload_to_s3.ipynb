# 00 - Data Upload to S3

This notebook uploads the Yelp parquet datasets to Amazon S3.

Datasets:

- yelp_reviews_2019.parquet
- yelp_reviews_2020-2022.parquet

These files will be used throughout the project for:

- AWS Glue
- Amazon Athena
- Feature Engineering
- Model Training
- Production Evaluation

The datasets are uploaded once and reused by all subsequent notebooks.

```python
import boto3

session = boto3.Session(region_name="us-east-1")
s3 = session.client("s3")

print("AWS connection successful")
print(session.region_name)
```

```python
import boto3

REGION = "us-east-1"
SOURCE_BUCKET = "sagemaker-us-east-1-555419874521"

local_2019 = "/home/ec2-user/yelp_data_upload/yelp_reviews_2019.parquet"
local_2020_2022 = "/home/ec2-user/yelp_data_upload/yelp_reviews_2020-2022.parquet"

s3_key_2019 = "data/2019/yelp_reviews_2019.parquet"
s3_key_2020_2022 = "data/2020_2022/yelp_reviews_2020-2022.parquet"

s3 = boto3.client("s3", region_name=REGION)

print("Bucket:", SOURCE_BUCKET)
```

```python
s3.upload_file(local_2019, SOURCE_BUCKET, s3_key_2019)
print("Uploaded 2019 data")

s3.upload_file(local_2020_2022, SOURCE_BUCKET, s3_key_2020_2022)
print("Uploaded 2020-2022 data")
```

```python
for prefix in ["data/2019/", "data/2020_2022/"]:
    print(f"\n--- {prefix} ---")
    response = s3.list_objects_v2(Bucket=SOURCE_BUCKET, Prefix=prefix)

    for obj in response.get("Contents", []):
        size_mb = obj["Size"] / (1024 ** 2)
        print(obj["Key"], f"{size_mb:.1f} MB")
```
